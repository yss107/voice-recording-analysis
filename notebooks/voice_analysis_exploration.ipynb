## 2. Sample Data Generation

Since we don't have access to the actual voice dataset, we'll generate realistic sample data that simulates the characteristics of male and female voices:

````python
def create_realistic_voice_data():
    """
    Create sample voice features based on known acoustic differences between male and female voices.
    """
    # Simulate features for male voices (typically lower frequencies)
    n_male = 150
    male_features = []
    
    for i in range(n_male):
        features = {
            'mean_frequency': np.random.normal(150, 30),  # Lower mean frequency
            'spectral_entropy': np.random.normal(6.5, 1.0),
            'mode_frequency': np.random.normal(120, 25),
            'f0_mean': np.random.normal(125, 20),  # Fundamental frequency
            'f0_std': np.random.normal(15, 5),
            'f0_range': np.random.normal(80, 20),
            'voiced_fraction': np.random.normal(0.65, 0.1),
            'formant_f1': np.random.normal(500, 50),  # First formant
            'formant_f2': np.random.normal(1200, 100),  # Second formant
            'formant_f3': np.random.normal(2400, 150),  # Third formant
            'spectral_centroid_mean': np.random.normal(1200, 200),
            'spectral_rolloff_mean': np.random.normal(3000, 400),
            'spectral_bandwidth_mean': np.random.normal(800, 100),
            'zero_crossing_rate_mean': np.random.normal(0.05, 0.01),
            'rms_energy_mean': np.random.normal(0.02, 0.005),
            'duration': np.random.normal(3.0, 1.0),
            'tempo': np.random.normal(120, 20)
        }
        male_features.append(features)
    
    # Simulate features for female voices (typically higher frequencies)
    n_female = 150
    female_features = []
    
    for i in range(n_female):
        features = {
            'mean_frequency': np.random.normal(220, 40),  # Higher mean frequency
            'spectral_entropy': np.random.normal(7.0, 1.2),
            'mode_frequency': np.random.normal(200, 35),
            'f0_mean': np.random.normal(210, 30),  # Higher fundamental frequency
            'f0_std': np.random.normal(20, 6),
            'f0_range': np.random.normal(100, 25),
            'voiced_fraction': np.random.normal(0.70, 0.1),
            'formant_f1': np.random.normal(600, 60),  # Higher formants
            'formant_f2': np.random.normal(1400, 120),
            'formant_f3': np.random.normal(2800, 180),
            'spectral_centroid_mean': np.random.normal(1500, 250),
            'spectral_rolloff_mean': np.random.normal(3500, 500),
            'spectral_bandwidth_mean': np.random.normal(900, 120),
            'zero_crossing_rate_mean': np.random.normal(0.07, 0.015),
            'rms_energy_mean': np.random.normal(0.018, 0.004),
            'duration': np.random.normal(3.2, 1.2),
            'tempo': np.random.normal(125, 22)
        }
        female_features.append(features)
    
    # Combine features and labels
    all_features = male_features + female_features
    labels = ['male'] * n_male + ['female'] * n_female
    
    return all_features, labels

# Generate sample data
print("Generating realistic voice feature data...")
features_list, labels = create_realistic_voice_data()

print(f"Generated {len(features_list)} samples")
print(f"Male samples: {labels.count('male')}")
print(f"Female samples: {labels.count('female')}")
print(f"Features per sample: {len(features_list[0])}")
````

## 3. Data Exploration and Analysis

````python
# Initialize data analyzer
analyzer = VoiceDataAnalyzer()

# Create DataFrame for analysis
df = analyzer.create_features_dataframe(features_list, labels)

print("Dataset Overview:")
print(f"Shape: {df.shape}")
print(f"\nColumns: {list(df.columns)}")
print(f"\nFirst few rows:")
df.head()
````

````python
# Generate summary statistics
summary_stats = analyzer.generate_summary_statistics(df)
print("Summary Statistics by Gender:")
summary_stats.head(10)
````

````python
# Plot feature distributions
print("Plotting feature distributions by gender...")
key_features = ['mean_frequency', 'spectral_entropy', 'mode_frequency', 
                'f0_mean', 'formant_f1', 'formant_f2']
analyzer.plot_feature_distributions(df, features=key_features)
````

````python
# Create box plots for key features
print("Creating box plots for key features...")
analyzer.plot_feature_boxplots(df, features=key_features)
````

````python
# Plot correlation matrix
print("Plotting feature correlation matrix...")
analyzer.plot_correlation_matrix(df)
````

## 4. Principal Component Analysis

````python
# Perform PCA analysis
print("Performing PCA analysis...")
analyzer.plot_pca_analysis(df)
````

## 5. Machine Learning Model Training

````python
# Initialize classifier
classifier = GenderClassifier(random_state=42)

# Prepare features for training
print("Preparing features for machine learning...")
X, feature_names = classifier.prepare_features(features_list)
y = np.array(labels)

print(f"Feature matrix shape: {X.shape}")
print(f"Number of features: {len(feature_names)}")
print(f"Feature names: {feature_names[:10]}...")  # Show first 10
print(f"Label distribution: {np.unique(y, return_counts=True)}")
````

````python
# Train multiple models
print("Training multiple machine learning models...")
results = classifier.train_all_models(X, y)

# Create model comparison report
comparison_df = classifier.create_model_comparison_report(results)
print("\nModel Performance Comparison:")
comparison_df
````

````python
# Plot model performance comparison
analyzer.plot_model_performance(results)
````

## 6. Feature Importance Analysis

````python
# Get and plot feature importance
try:
    importance = classifier.get_feature_importance()
    
    print("Feature Importance:")
    for i, (feature, score) in enumerate(list(importance.items())[:10]):
        print(f"{i+1:2d}. {feature.replace('_', ' ').title()}: {score:.4f}")
    
    # Plot feature importance
    analyzer.plot_feature_importance(importance, top_n=15)
    
except Exception as e:
    print(f"Could not extract feature importance: {e}")
````

## 7. Model Evaluation

````python
# Evaluate the best model
evaluation = classifier.evaluate_model(classifier.best_model, X, y)

print(f"Best Model Accuracy: {evaluation['accuracy']:.4f}")
print(f"\nConfusion Matrix:")
print(evaluation['confusion_matrix'])

# Plot confusion matrix
analyzer.plot_confusion_matrix(
    y, evaluation['predictions'], 
    labels=['female', 'male']
)
````

````python
# Detailed classification report
print("Detailed Classification Report:")
class_report = evaluation['classification_report']
for label, metrics in class_report.items():
    if isinstance(metrics, dict):
        print(f"\n{label.title()}:")
        for metric, value in metrics.items():
            if isinstance(value, (int, float)):
                print(f"  {metric}: {value:.4f}")
````

## 8. Making Predictions

````python
# Test prediction on sample data
sample_features = features_list[0]  # Use first sample
actual_label = labels[0]

prediction_result = classifier.predict(sample_features)

print("Sample Prediction:")
print(f"Actual gender: {actual_label}")
print(f"Predicted gender: {prediction_result['prediction']}")

if 'confidence' in prediction_result:
    print(f"Confidence: {prediction_result['confidence']:.4f}")

if 'probabilities' in prediction_result:
    print("Probabilities:")
    for gender, prob in prediction_result['probabilities'].items():
        print(f"  {gender}: {prob:.4f}")
````

## 9. Hyperparameter Tuning (Optional)

````python
# Perform hyperparameter tuning for the Random Forest model
print("Performing hyperparameter tuning for Random Forest...")
tuning_results = classifier.hyperparameter_tuning(X, y, 'random_forest')

print(f"\nBest parameters: {tuning_results['best_params']}")
print(f"Best cross-validation score: {tuning_results['best_score']:.4f}")

# Update the best model with tuned parameters
classifier.best_model = tuning_results['best_model']
print("\nModel updated with best parameters.")
````

## 10. Model Persistence

````python
# Save the trained model
import os

# Create models directory
models_dir = '../models'
os.makedirs(models_dir, exist_ok=True)

# Save model
model_path = os.path.join(models_dir, 'gender_classifier.joblib')
classifier.save_model(model_path)

print(f"Model saved to: {model_path}")

# Test loading the model
new_classifier = GenderClassifier()
new_classifier.load_model(model_path)
print("Model loaded successfully!")

# Verify the loaded model works
test_prediction = new_classifier.predict(sample_features)
print(f"Test prediction with loaded model: {test_prediction['prediction']}")
````
